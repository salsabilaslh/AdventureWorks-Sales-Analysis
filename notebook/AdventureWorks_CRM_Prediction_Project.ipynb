```python
import pandas as pd

file_path = "/content/AdventureWorks Sales.xlsx"

excel_file = pd.ExcelFile(file_path)

print(excel_file.sheet_names)
```

```python
sales = pd.read_excel(file_path, sheet_name='Sales_data')
customers = pd.read_excel(file_path, sheet_name='Customer_data')
products = pd.read_excel(file_path, sheet_name='Product_data')
territory = pd.read_excel(file_path, sheet_name='Sales Territory_data')
dates = pd.read_excel(file_path, sheet_name='Date_data')
```

```python
print(sales.head())
```

```python
# Merge sales + customers
df = sales.merge(customers, on='CustomerKey', how='left')

# Merge products
df = df.merge(products, on='ProductKey', how='left')

# Merge territory
df = df.merge(territory, on='SalesTerritoryKey', how='left')

print(df.shape)
df.head()
```

```python
# Check missing values
df.isnull().sum()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df['Color'] = df['Color'].fillna('Unknown')
```

```python
df = df.dropna(subset=['ShipDateKey'])
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
print(df.shape)
```

```python
df.dtypes
```

```python
df['OrderDateKey'] = pd.to_datetime(df['OrderDateKey'], format='%Y%m%d')
df['DueDateKey'] = pd.to_datetime(df['DueDateKey'], format='%Y%m%d')
df['ShipDateKey'] = pd.to_datetime(df['ShipDateKey'].astype(int), format='%Y%m%d')
```

```python
df.dtypes
```

```python
# Remove invalid sales values
df = df[df['Sales Amount'] > 0]

# Remove invalid quantity
df = df[df['Order Quantity'] > 0]

# Remove invalid price
df = df[df['Unit Price'] > 0]
```

```python
print(df.shape)
```

```python
df.to_csv("clean_adventureworks.csv", index=False)
```
